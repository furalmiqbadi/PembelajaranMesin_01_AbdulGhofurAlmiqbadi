# Langkah 0 - Persiapan

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report
```

```python
df = pd.read_csv('/content/wbc.csv')
df.head()
```

# Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.

```python
df.columns
```

```python
unusable_columns = ['id', 'Unnamed: 32']
df_cleaned = df.drop(columns=unusable_columns)

df_cleaned.columns
```

# Lakukan proses encoding pada kolom "diagnosis".



```python
# diagnosis itu dependent jadi pakai Label Encoding
le = LabelEncoder()
df_cleaned['diagnosis'] = le.fit_transform(df_cleaned['diagnosis'])

df_cleaned.head()
```

# Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik.



```python
std = StandardScaler()
num_cols = df_cleaned
std_num = [col for col in num_cols if col != 'diagnosis']
df_cleaned[std_num] = std.fit_transform(df_cleaned[std_num])


df_cleaned.head()
```

# Lakukan proses seleksi fitur. Anda dapat menggunakan SelectKBest.



```python
#pisahkan fitur (X) dan target (y)
x = df_cleaned.drop('diagnosis', axis=1)
y = df_cleaned['diagnosis']

#pilih fitur terbaik
selector = SelectKBest(score_func=f_classif)
x_best = selector.fit_transform(x, y)

fitur_terpilih = x.columns[selector.get_support()]
print(fitur_terpilih)
```

# Lakukan proses pengujian dengan model Logistic Regression seperti pada praktikum 1.

```python
#bagi data menjadi training dan testing (80:20)
x_train, x_test, y_train, y_test = train_test_split(x_best, y, test_size=0.2, random_state=42)

#buat Pipeline sederhana menggunakan LogisticRegression
pelatihan = Pipeline([
    ('model', LogisticRegression(solver='liblinear'))
])

#latih model
pelatihan.fit(x_train, y_train)

#prediksi dan cek akurasi
y_pred = pelatihan.predict(x_test)
akurasi = accuracy_score(y_test, y_pred)

print(f"Akurasi Model: {akurasi * 100:.2f}%")
print(classification_report(y_test, y_pred))
```

# Berdasarkan hasil analisa Anda, berapa jumlah fitur terbaik yang dapat digunakan? Apa saja fitur tersebut?

```python
print(f"Jumlah Fitur: \n{len(fitur_terpilih)}")
print(f"Daftar fitur: \n{list(fitur_terpilih)}")
```
