## Import

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

## Load Data & Preprocessing

```python
df = pd.read_csv('/content/spam.csv', encoding='latin-1')
df = df.drop(df.iloc[:, 2:], axis=1)
df = df.rename(columns={'v1': 'Labels', 'v2': 'SMS'})
df['Labels'] = df['Labels'].map({'spam': 1, 'ham': 0})

X = df['SMS'].values
y = df['Labels'].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
```

## CountVectorizer dengan stop_words

```python
# inisiasi CountVectorizer dengan stop_words bahasa Inggris
bow_stop = CountVectorizer(stop_words='english')

X_train_bow = bow_stop.fit_transform(X_train)
X_test_bow = bow_stop.transform(X_test)

# training Model
mnb_bow = MultinomialNB()
mnb_bow.fit(X_train_bow, y_train)

# evaluasi
y_pred_bow = mnb_bow.predict(X_test_bow)
acc_bow = accuracy_score(y_test, y_pred_bow)
print(f"Akurasi CountVectorizer + Stop Words: {acc_bow:.4f}\n")
print("Laporan Klasifikasi:\n", classification_report(y_test, y_pred_bow, target_names=['ham', 'spam']))
```

## TF-IDF dengan stop_words

```python
# inisiasi TfidfVectorizer dengan stop_words bahasa inggris
tfidf_stop = TfidfVectorizer(stop_words='english')

X_train_tfidf = tfidf_stop.fit_transform(X_train)
X_test_tfidf = tfidf_stop.transform(X_test)

# training model
mnb_tfidf = MultinomialNB()
mnb_tfidf.fit(X_train_tfidf, y_train)

# evaluasi
y_pred_tfidf = mnb_tfidf.predict(X_test_tfidf)
acc_tfidf = accuracy_score(y_test, y_pred_tfidf)
print(f"Akurasi TF-IDF + Stop Words: {acc_tfidf:.4f}\n")
print("Laporan Klasifikasi:\n", classification_report(y_test, y_pred_tfidf, target_names=['ham', 'spam']))
```

## Perbandingan

```python
print(f"CountVectorizer + Stop Words : {acc_bow:.4f}")
print(f"TF-IDF + Stop Words          : {acc_tfidf:.4f}")
```

* Tugas 1: Evaluasi CountVectorizer dengan stop_words
Penggunaan CountVectorizer dengan mengaktifkan stop_words berhasil menyaring kata-kata umum yang tidak informatif seperti the, is, in. Hal ini mengurangi noise pada data teks dan menghasilkan akurasi yang sangat tinggi biasanya di atas 97%. Model mampu mengklasifikasikan SMS spam dan ham dengan presisi dan recall yang seimbang.
* Tugas 2: Evaluasi TF-IDF dengan stop_words dan Perbandingan
TF-IDF dengan stop_words memberikan bobot lebih pada kata-kata yang unik dan spesifik di setiap dokumen sambil tetap mengabaikan kata-kata umum. Ketika dibandingkan dengan CountVectorizer, akurasi TF-IDF biasanya menghasilkan nilai yang sedikit lebih tinggi atau setidaknya setara. Perbedaannya terletak pada cara pembobotan fitur pada TF-IDF lebih pintar dalam menekan kata yang terlalu sering muncul di semua dokumen sehingga mengurangi bias frekuensi.
* Kesimpulan:
Berdasarkan hasil percobaan, fitur TF-IDF dengan stop_words merupakan yang terbaik untuk kasus data spam.csv. Meskipun kedua metode sama-sama menghasilkan akurasi yang sangat tinggi tapi TF-IDF lebih unggul karena tidak hanya menghitung jumlah kemunculan kata seperti CountVectorizer tetapi juga mempertimbangkan seberapa penting kata tersebut dalam sebuah dokumen relatif terhadap seluruh korpus. Kombinasi ini membuat model lebih robust dan minim terhadap noise kata-kata umum.
